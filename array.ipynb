```python
def average_Mark(register):
    if len(register)==0:
        return 0
    total=0
    for s in register:
        total=total+s["Mark"]
    return total /len(register)
    
def count_present(register):
     count=0
     for s in register:
         if s["attendance"]=="P":
             count += 1
         return count
```
